```python
# import pandas as pd
import polars as pl
```

```python
days = list(range(23,26+1))

# Large files, download them from MarineCadastre with given dates, linked at bottom of the article
path = '../data/geodata/ais/AIS_2021_07_{}/AIS_2021_07_{}.csv'
for i, day in enumerate(days):
    ais = pl.read_csv(path.format(day, day))
    ama_new = ais.filter(ais['VesselName']=='ARTHUR M ANDERSON')
    if i==0:
        ama = ama_new
    else:
        ama.vstack(ama_new, in_place=True)
```

```python
ama.write_csv('../data/geodata/arthurmanderson_2021_07_{}to{}.csv'.format(min(days), max(days)))
# Then converted to geojson in QGIS
```
